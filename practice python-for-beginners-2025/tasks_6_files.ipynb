### Задача 1

Создайте самостоятельно на компьютере файл сосстоящий из 1 строки, в которой будет 10 слов. Прочитайте этот файл с помощью кода, возьмите строку из файла и постройте лесенку. Т.е., сначала 1 слово, потом первые два слова, затем первые три слова и так до конца (используйте цикл).

```python
path ='C:\\Users\\user\\Desktop\\ex.txt' 
m=str()
with open(path, 'r', encoding='utf-8') as output:
    f = output.read()
    print(f)
    f=f.split(' ')
    for i in range(len(f)):
        m=m + str(f[i]) + ' '
        print(m)
```

### Задача 17
Напишите код, который:
- Создаёт в текущей директории папку с именем "results" (если она ещё не существует).
- Находит в текущей папке все файлы с расширением ".txt".  
- Для каждого такого файла считает количество строк.  
- Создаёт новый файл "summary.txt" в папке "results", куда записывает имена файлов и количество строк в них.  

```python
#os.mkdir('results')
line_count=[]
line_count_in_files={}
l=[f for f in os.listdir('results') if f.endswith('.txt')]

for i, file_name in enumerate(l):
    with open(l[i]) as f:
        x=0
        while f.readline():
            x+=1
            line_count.append(x+1)
        line_count_in_files.setdefault(file_name, x)  
with open('summary.txt', 'w') as summ:
        summ.write(str(line_count_in_files))
print(line_count_in_files)
```
