## Задание 4 Конфликт и его решение

Создаем новую ветку `conflict`. Открываем файл `file_for_conflicts.txt`

удаляем строку в файле которую нужно удалить → `Commit`.

`Branch → Switch to branch → main`, в этом же файле теперь делаем добавляем что-то там, где есть символ `<>`  → `Commit`.

Далее идем в `Branch → Merge into current branch` → выбераем `conflict` → `main`

Разрешаем конфликт:
### Как разрешать конфликт вручную:
```
Lorem ipsum dolor sit amet, consectetur adipiscing elit <something>, sed do eiusmod tempor incididunt ut labore et dolore magna aliqua.
Ut enim ad minim veniam, quis nostrud exercitation ullamco laboris nisi ut aliquip ex ea commodo consequat.
<Эту строчку нужно удалить, или нет? Да, удалите её!!!!>
Duis aute <something_new> irure dolor in reprehenderit in voluptate velit <Лишнее слово!> esse cillum dolore eu fugiat nulla pariatur.
<<<<<<< HEAD
Excepteur sint occaecat cupidatat non <Not word> proident, sunt in culp
<Thing>
=======
Excepteur sint occaecat cupidatat non <Word> proident, sunt in culp
<Nothing>
>>>>>>> conflict
>>>>>>> parent of 02e5786 (удалена строчка в конфликте)
```

Это означает, что версия из HEAD (вашей ветки *master*, то что вы извлекли из удалённого репозитория для слияния) — это верхняя часть блока (всё, что над `=======`), а версия из вашей ветки *conflict* представлена в нижней части. 
Чтобы разрешить конфликт, придётся выбрать один из вариантов, либо объединить содержимое по-своему. Например, вы можете разрешить конфликт, заменив весь блок следующим:

Какой-то текст<br>
Ещё немного какого-то текста<br>

В этом разрешении есть немного от каждой части, а строки `<<<<<<<`, `=======` и `>>>>>>>` должны быть полностью удалены. Разрешив каждый конфликт, чтобы отметить конфликт как решённый. Добавление файла в индекс означает для Git, что все конфликты в нём исправлены.
