## Задача 13

Рассчитайте площадь треугольника со сторонами 3, 4, 5 по формуле Герона:

$$ S = \sqrt{p \cdot (p-a) \cdot (p-b) \cdot (p-c)} $$

где $ p = \frac{a + b + c}{2} $ — полупериметр треугольника.


```python
# your code here
import math
a=3
b=4
c=5
p=(a+b+c)/2
s=math.sqrt(p*(p-a)*(p-b)*(p-c))
print(s)
```

```python
# your code here
x=int(input('Введите время митинов, часов'))
y=int(input('Введите ввремя сна, часов'))
z=int(input('Введите время отдыха, минут'))
z/=60
disser_time=24-x-y-z-7
disser_time_hours=round(disser_time)
disser_time_minutes=round((float(disser_time)-int(disser_time))*60)
if x+y+z > 24-7:
    print('Много сна, митингов или отдыха')
else:
    print(disser_time_hours, ' часов ', disser_time_minutes, ' минут')

```
