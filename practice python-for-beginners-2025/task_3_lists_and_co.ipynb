### Задача 14

Вам даны 2 строки:

`sampleA: lacZ, recA, rpoB, dnaK`

`sampleB: recA, dnaK, groEL, rpoB`

Для каждой строки:

- отделить имя образца (часть до :) и список генов (после :).

- список генов превратить в список строк без пробелов по краям у каждого элемента.

Собрать структуру данных:

- кортеж для каждого образца вида (sample_name, genes_list, genes_set)

Посчитать и вывести:

- множество общих генов
- множество уникальных генов каждого образца
- общий список генов в алфавитном порядке (без повторов)

Вывести итог одной строкой в формате:
`A=<имяA>; B=<имяB>; common=<...>; A_only=<...>; B_only=<...>`

отделяем одно от другого. Разделяем гены через сплит и создание списка. Соединяем список образцов и список генов. Преобразуем список в кортеж

````python
# your code here
s1='sampleA: lacZ, recA, rpoB, dnaK'
s2='sampleB: recA, dnaK, groEL, rpoB'
nameA, genesA = s1.split(':')
nameB, genesB = s2.split(':')
genesA=genesA.strip().split(', ')
genesB=genesB.strip().split(', ')
tupleA=(nameA, genesA, set(genesA))
tupleB=(nameB, genesB, set(genesB))
general_genes=sorted(tupleA[2]|tupleB[2])
common=tupleA[2]&tupleB[2]
A_only= tupleA[2]-tupleB[2]
B_only= tupleB[2]-tupleA[2]

print(f'A={nameA}; B={nameB}; common={common}; A_only={A_only}; B_only={B_only}; general genes={general_genes}')

````

### Задача 15

Вам дана строка:
`P1|OS=Homo sapiens|GN=HBA1|TAG=heme,oxygen,blood || P2|OS=Homo sapiens|GN=HBB|TAG=oxygen,blood,globin`

Сделать следующее:
- Разделить строку на две части (два белка) по || и удалить концевые пробелы у каждой части.

Для каждого белка:

- получить protein_id (до первого |)
- получить значение OS
- получить значение GN
- получить строку тегов после TAG и превратить её в список тегов
- получить только уникальные теги

Собрать по каждому белку кортеж вида:
`(protein_id, OS, GN, tags_list, tags_set)`

Посчитать и вывести:

- только общие теги
- все теги без повторов и отсортированный список из них

````python
# your code here
s='P1|OS=Homo sapiens|GN=HBA1|TAG=heme,oxygen,blood || P2|OS=Homo sapiens|GN=HBB|TAG=oxygen,blood,globin'
p1, p2 = s.split('||')
p1=p1.strip()
p2=p2.strip()
parts1=p1.split('|')
parts2=p2.split('|')

OS1=parts1[1].split('=')
GN1=parts1[2].split('=')
tags_list1=parts1[3].split('=')
tags_list1=tags_list1[1].split(',')

OS2=parts2[1].split('=')
GN2=parts2[2].split('=')
tags_list2=parts2[3].split('=')
tags_list2=tags_list2[1].split(',')

cort1= (parts1[0], OS1[1], GN1[1], tags_list1, set(tags_list1))
cort2= (parts2[0], OS2[1], GN2[1], tags_list2, set(tags_list2))
common=cort1[4]&cort2[4]
all=sorted(cort1[4]|cort2[4])
print(f'common tags = {common}, all sorted genes = {all}')

````
