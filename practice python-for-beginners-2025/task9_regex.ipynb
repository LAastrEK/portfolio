```python
s = '94НН03 С006Щ3НN3 ПОК43Ы8437, К4КИ3 У9И8И73ЛЬНЫ3 83ЩИ М0Ж37 93Л47Ь Н4Ш Р43УМ!'\
    '8П3Ч47ЛЯЮЩИ3 83ЩИ! СН4Ч4Л4 Э70 6ЫЛ0 ТРУДНО, Н0 С3ЙЧ4С Н4 Э70Й С7Р0К3'\
    '84Ш РАЗУМ ЧИ7437 Э70 4870М47ИЧ3СКИ, Н3 349УМЫ84ЯСЬ 06 Э70М. ГОРДИСЬ! Л4ШЬ 0ПР393Л3ННЫ3'\
    'ЛЮ9Н М0ГУ7 ПР0ЧИТАТЬ Э70'
```

Задача 2. Список содежит наименования учереждений. Используя регулярные выражения создайте аббревиатуры для каждого из наименований. Результат внесите в словарь, где ключ аббревиатура, значение полное наименование.

```python
institutions = ['Московский главный архив Министерства иностранных дел',\
                'Московский институт истории, философии и литературы',\
                'Отдел истории первобытной культуры',\
                'Нижегородский государственный художественный музей',\
                'Детский городской многопрофильный клинический центр высоких медицинских технологий']
```

```python
# paste your code here
res = {}
for i in institutions:
    abr = ''.join(re.findall(r'\b[А-Яа-я]', i)).upper()
    res[abr] = i

print(res)
```

```python
letter = 'Здравствуй, дорогой друг.'\
         'Спешу уведомить тебя, что наша завтрашняя встреча переносится '\
         'с 11:30, на 18:40. Дело в том, что с 13:05 и до 17:59, я буду  '\
         'занят. О! Еще я не правильно указал название бара, 42:03, а не 24:30. '\
         'Будь внимателен! Кстати спасибо за рецепт борща. Пропорция 110:20 идеальная!'
```

```python
# paste your code here
x=re.sub(r'\b([01]\d|2[0-3]):[0-5]\d\b', 'TIME', letter)
x
```
