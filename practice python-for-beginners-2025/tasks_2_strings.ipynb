$$ GC\% = \frac{G + C}{L} \times 100\% $$

где G и С - количество гуанина и цитозина, соответственно; L - длина последовательности.


````python
seq = 'ATGCGTACGATCGA'
# your code here
g=seq.count('G')
c=seq.count('C')
l=len(seq)
GC_perc=(g+c)/l*100
print(GC_perc, '%')
````
