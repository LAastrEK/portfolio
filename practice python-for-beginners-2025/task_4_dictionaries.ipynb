### Задача 2

````python
# Решение задача
s= input()
a = s.count("A")
c = s.count("C")
g = s.count("G")
t = s.count("T")
gc = ((g + c) / len(s)) * 100
result = {"length": len(s), "counts": {"A": a, "C": c, "G": g, "T": t}, "gc": round(gc, 2)}
print(result)
````
